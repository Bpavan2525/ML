## Q1. What is Simple Linear Regression (SLR)? Explain its purpose.
  - Simple Linear Regression (SLR) is a statistical and machine learning technique used to model the relationship between one independent variable and one dependent variable using a straight-line equation. It assumes that the dependent variable changes linearly with respect to the independent variable and represents this relationship using the equation (Y = mX + c), where (Y) is the dependent variable, (X) is the independent variable, (m) is the slope, and (c) is the intercept.
  - The main purpose of Simple Linear Regression is to understand how one variable influences another and to make predictions based on that relationship. It helps identify trends, measure the strength and direction of association, and estimate future values of the dependent variable using known values of the independent variable.

## Q2. What are the key assumptions of Simple Linear Regression?
  - The key assumptions of Simple Linear Regression ensure that the model produces reliable and meaningful results. One important assumption is linearity, which means there should be a straight-line relationship between the independent variable and the dependent variable. Another assumption is independence, where each observation should be independent of the others.
  - The model also assumes homoscedasticity, meaning the variance of the errors or residuals remains constant across all values of the independent variable. Additionally, the residuals are assumed to be normally distributed, which is important for statistical inference and hypothesis testing. Finally, there should be no significant outliers that could strongly influence the regression line and distort the model's accuracy.

## Q3. Write the mathematical equation for a simple linear regression model and explain each term.
  - The mathematical equation for a Simple Linear Regression model is:
    - Y = mX + c
      - In this equation, (Y) represents the dependent variable, which is the value we want to predict. The variable (X) is the independent variable, which is used to explain or predict changes in (Y). The coefficient (m) is called the slope of the regression line and indicates how much (Y) changes for a one-unit increase in (X). The term (c) is the intercept, which represents the value of (Y) when (X) is equal to zero. Together, these components form a straight-line model that describes the relationship between the two variables.


## Q4. Provide a real-world example where simple linear regression can be applied.

In this case, the size of the house in square feet acts as the independent variable, while the house price is the dependent variable. As the size of the house increases, the price generally increases as well. Simple linear regression can be used to model this relationship and estimate the expected price of a house based on its size. ## Q5. What is the method of least squares in linear regression?
  - The method of least squares in linear regression is a technique used to find the best-fitting line that represents the relationship between the independent and dependent variables. It works by minimizing the sum of the squared differences between the actual observed values and the predicted values given by the regression line. These differences are called residuals or errors.
  - By squaring the errors and adding them together, the method ensures that larger errors have a greater impact and that positive and negative errors do not cancel each other out. The regression line obtained through the least squares method is considered the optimal line because it produces the smallest possible total prediction error.

## Q6. What is Logistic Regression? How does it differ from Linear Regression?
  - Logistic Regression is a supervised machine learning algorithm used for classification problems, where the dependent variable represents categories such as yes or no, true or false, or 0 and 1. Instead of predicting continuous values, logistic regression predicts the probability that an observation belongs to a particular class. It uses a logistic or sigmoid function to convert the output into values between 0 and 1, which can then be interpreted as probabilities.
  - The main difference between Logistic Regression and Linear Regression lies in their purpose and output. Linear regression is used for predicting continuous numerical values and models the relationship using a straight-line equation. Logistic regression, on the other hand, is used for classification tasks and produces probability-based outputs that are mapped to categories. While linear regression uses a linear function, logistic regression applies a sigmoid function to handle classification problems effectively.

## Q7. Name and briefly describe three common evaluation metrics for regression models.
  - Three common evaluation metrics for regression models are Mean Absolute Error (MAE), Mean Squared Error (MSE), and the Coefficient of Determination (( R^2 )). Mean Absolute Error measures the average absolute difference between the actual values and the predicted values, giving a simple interpretation of prediction error without considering direction. Mean Squared Error calculates the average of the squared differences between actual and predicted values, which penalizes larger errors more heavily and is useful for detecting significant prediction mistakes. The Coefficient of Determination, or ( R^2 ), measures how well the model explains the variation in the dependent variable, indicating the proportion of variance captured by the regression model.

## Q8. What is the purpose of the R-squared metric in regression analysis?
  - The purpose of the R-squared (( R^2 )) metric in regression analysis is to measure how well a regression model explains the variation in the dependent variable. It indicates the proportion of the total variability in the outcome that is accounted for by the independent variables in the model. The value of ( R^2 ) ranges from 0 to 1, where a value closer to 1 means the model explains a large portion of the data variation, and a value closer to 0 means the model explains very little.
  - R-squared helps evaluate the goodness of fit of a regression model by showing how effectively the model captures the relationship between variables. However, it should be used along with other evaluation measures to fully assess model performance.

## Q9. Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.

```python
# Q9. Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 5, 4, 5])

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

## Q10. How do you interpret the coefficients in a simple linear regression model?
  - In a simple linear regression model, the coefficients describe how the dependent variable changes with respect to the independent variable. The slope coefficient represents the amount of change in the dependent variable for a one-unit increase in the independent variable. If the slope is positive, it means the dependent variable increases as the independent variable increases, and if it is negative, the dependent variable decreases as the independent variable increases.
  - The intercept represents the predicted value of the dependent variable when the independent variable is equal to zero. Together, these coefficients help explain the relationship between variables and allow the model to make predictions based on input values.
